```python
base_path =  "/Volumes/workspace/default/volumewd36"
raw_path = f"{base_path}/raw"
processed_path = f"{base_path}/processed"
archive_path = f"{base_path}/archive"

```

```python
%sql
CREATE VOLUME IF NOT EXISTS workspace.default.volumewd36;
```

```python
# Create required directories in Databricks using dbutils
dbutils.fs.mkdirs(raw_path)
dbutils.fs.mkdirs(processed_path)
dbutils.fs.mkdirs(archive_path)

print("Directories created successfully!")
```

```python
sample_data = """patient_id,patient_name,age,gender
1,John Doe,68,M
2,Jane Smith,54,F
"""
```

```python
# File path inside raw folder
file_path = f"{raw_path}/healthcare_sample.csv"

# Write file content using dbutils
dbutils.fs.put(file_path, sample_data, overwrite=True)

print(f"Healthcare sample file created at: {file_path}")
```

```python
display(dbutils.fs.ls(raw_path))
```

```python
source_file = f"{raw_path}/healthcare_sample.csv"
target_file = f"{processed_path}/healthcare_sample.csv"

# Copy file (overwrite = True)
dbutils.fs.cp(source_file, target_file, recurse=False)

print("File copied successfully!")
```

```python
dbutils.widgets.dropdown(
    name="environment",
    defaultValue="dev",
    choices=["dev", "qa", "prod"],
    label="Select Environment"
)
```

```python
dbutils.widgets.text(
    name="owner",
    defaultValue="",
    label="Owner Name"
)
```

```python
# Read values from widgets
environment = dbutils.widgets.get("environment")
owner = dbutils.widgets.get("owner")

# Print the values
print(f"Selected Environment: {environment}")
print(f"Owner Name: {owner}")
```

```python
# File paths
source_file = f"{processed_path}/healthcare_sample.csv"
target_file = f"{archive_path}/healthcare_sample.csv"

# Move the file to archive
dbutils.fs.mv(source_file, target_file, recurse=False)

print("Processed file moved to archive successfully!")
```
