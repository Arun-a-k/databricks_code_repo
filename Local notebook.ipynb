```python
# %python (not needed explicitly in Python cells)
file_path = "/Volumes/workspace/default/usage_metrics/mobile_os_usage.csv"

df1 = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load(file_path)

df1.show(5)
df1.printSchema()
```

Task7: Write the above dataframe df1 data into a databricks table called 'default.mobile_os_usage' using pyspark magic command %python


```python
%python
# Rename columns to remove spaces and invalid characters
df1_clean = df1.withColumnRenamed(
    "Mobile Operating System", "mobile_operating_system"
).withColumnRenamed(
    "Percent of Usage", "percent_of_usage"
)

df1_clean.write.mode("overwrite").saveAsTable("default.mobile_os_usage")

print("Table created successfully: default.mobile_os_usage")
```

Task8: Write sql query to display the data loaded into the table 'default.mobile_os_usage' using the pyspark magic command %python


```python
# Query the table using Spark SQL
result_df = spark.sql("SELECT * FROM default.mobile_os_usage")

# Display the result
display(result_df)
```
